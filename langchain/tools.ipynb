```python
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3-32b")
```

```python
response = model.invoke("Write a poem about the power of AI in shaping the future of humanity.")
print(response)
```

```python
from langchain.tools import tool
from langchain.agents import create_agent

@tool
def get_weather(location:str)->str:
    """get the weather of the given location"""
    return f"it is sunny in {location} today"


agent= create_agent(
    model= "groq:qwen/qwen3-32b",
    tools = [get_weather],
    system_prompt = "You are a helpful assistant that can provide weather information."
)

```

```python
response = agent.invoke({"messages":[{"role":"user","content":"What is the weather in New York City today?"}]})
```
